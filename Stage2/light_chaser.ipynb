```python
#The program runs the LEDs in sequence from 0 to 3 and back again over and over.
#The speed can be controlled via switch 0. It pauses upon pressing BTN 0
#And stops when pressing BTN 3

from pynq.overlays.base import BaseOverlay
from time import sleep

base = BaseOverlay("base.bit")
```

```python
#Defining variables
delaySlow = 0.9
delayFast = 0.3
delay = 0
ledNumber = 0
upDown = 1
steps = 9

lightChaserOn = 1
```

```python
while(lightChaserOn == 1):  

    base.leds[ledNumber].on()
        
    for i in range(steps):
        sleep(delay/steps)
        while(base.buttons[0].read()==1):
            sleep(0.1)
            
        if(base.switches[0].read() == 0):
            delay = delaySlow
        else:
            delay = delayFast
            
        if(base.buttons[3].read() == 1):
            lightChaserOn = 0
            break

    base.leds[ledNumber].off()

    if(ledNumber == 3):
        upDown = -1
    elif(ledNumber == 0):
        upDown = 1
    ledNumber += upDown

        
        
for led in base.leds:
    led.off()
```
